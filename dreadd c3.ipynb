```python
#Cross Correlation Analysis
# <editor-fold desc="...">
f= dreadd_remote_m22
#60 sec before shock
original_x = f.dfz_signals['Region0R']
original_y = f.dfz_signals['Region1G']
# original_x = f.dfz_signals['Region0R'][0*30:60*30]
# original_y = f.dfz_signals['Region1G'][0*30:60*30]
shifted_versions = [(original_x, original_y)]


def correlation(x, y):
    shortest = min(x.shape[0], y.shape[0])
    return np.corrcoef(x.iloc[:shortest].values, y.iloc[:shortest].values)[0, 1]

# Function to plot time series and show the correlation
def plot_correlation(x, y, text):
    plt.subplots(figsize=(5, 3))
    x.plot(label="x", color='green')
    y.plot(label="y", color = 'red')
    plt.title(f"Correlation {text}: {correlation(x, y)}")
    plt.legend(loc="best")
    plt.show()
    
from scipy.signal import correlate
from scipy.signal import correlation_lags

from scipy.signal import correlate
import numpy as np
import matplotlib.pyplot as plt
def adjust_lag(lag, correlation,lags):
    if abs(lag)>5*30:
        int = max(range(len(correlation)), key=correlation.__getitem__)
        correlation[int]=0
        lag = lags[np.argmax(correlation)]
        return lag
# Function to calculate cross-correlation,
# extract the best matching shift and then shift
# one of the series appropriately.
def shift_for_maximum_correlation(x, y):
    correlation = correlate(x, y, mode="full")
    lags = correlation_lags(x.size, y.size, mode="full")
    lag = np.argmax(correlation) - (len(x) - 1)
    
    # Adjust large lags if needed
    while abs(lag)>5*30: #if more than this many frames lag, adjust it 
        lag = adjust_lag(lag,correlation,lags)
    
    print("Lag:", lag)


    if lag < 0:
        y = y.iloc[abs(lag):].reset_index(drop=True)
    else:
        x = x.iloc[lag:].reset_index(drop=True)
    return x, y

# Function to plot time series and show the correlation
def plot_correlation(x, y, text):
    plt.figure(figsize=(2, 1.5))
    plt.subplots(figsize=(2, 1.5))
    x.plot(label="x", color='green')
    y.plot(label="y", color='red')
    plt.title(f"Correlation {text}: {correlation(x, y)}")
    plt.legend(loc="best")
    plt.show()

# Show results without shifting
for x, y in shifted_versions:
    plot_correlation(x, y, "before shifting")

# Plot results after shifting
for x, y in shifted_versions:
    shifted_x, shifted_y = shift_for_maximum_correlation(x, y)
    plot_correlation(shifted_x, shifted_y, "after shifting")
# </editor-fold>
```
